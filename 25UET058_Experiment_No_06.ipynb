```python
tup = ()
print(tup)

# Using String
tup = ('Omkar', 'Patil')
print(tup)

# Using List
li = [1, 2, 3, 4]
print(tuple(li))

# Using Built-in Function
tup = tuple('DKTE')
print(tup)
```

```python
tup = (5, 'Welcome', 7.5, True, [1, 2, 3], {'key': 'value'})
print(tup)
```

```python
tup = tuple("DKTE")
print(tup[0])
print(tup[1:4])
print(tup[:3])

# Tuple unpacking
tup = ("All", "is", "well")

# This line unpack values of Tuple1
a, b, c = tup
print(a)
print(b)
print(c)
```

```python
tup1 = (0, 1, 2, 3)
tup2 = ('DKTE', 'ICHALKARNJI')
tup3 = tup1 + tup2
print(tup3)
```

```python
tup = tuple('ALLISWELL')
print(tup[1:])
print(tup[::-1])
print(tup[4:9])
```

```python
tup = (0, 1, 2, 3, 4)
print(tup)
del tup
```

```python
tup = (1, 2, 3, 4, 5)
a, *b, c = tup
print(a)
print(b)
print(c)
```
