```python
# REVERSE NUMBER
for i in range (10,0,-1):
     print(i)
```

```python
# EVEN NUMBER
for i in range(2, 102, 2):
    print(i)
```

```python
# ODD NUMBER
for i in range(1, 101, 2):
    print(i)
```

```python
# SUM OF GIVEN A NUMBERS
n = int(input("Enter a Number = "))
total_sum=n * (n+1)//2
print(f"The sum of all Number up to n :{total_sum}")
```

```python
# PRINT MULTIPLICATION OF A TABLE
n = int(input("Enter a Number = "))
for i in range(1, 11):
  print(f"{n}*{i}={n*i}")
```

```python
# PRINT CUBE OF A NOS. BETWEEN 1 TO 10
for i in range(1, 11):
  print(f"{i}*{i}*{i}={i*i*i}")
```

```python
# PRINT SQUARE OF A NOS. BETWEEN 1 TO 10
for i in range(1, 11):
    print(f"{i}*{i}={i*i}")

```

```python
sentence = input("Enter a sentence: ")
vowels = "aeiouAEIOU"
vowel_count = 0
for char in sentence:
    if char in vowels:
        vowel_count += 1

print("Number of vowels:", vowel_count)
```
