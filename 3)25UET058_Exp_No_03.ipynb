```python
# CHECK NATIONALITY AND ELIGIBILITY OF PERSON
age=int(input("Enter Your Age = "))
citizenship=input("Are you citizen of India? (yes/no): ")

if age >= 18 and citizenship =="yes":
  print("Eligible For Voting in India")
else :
  print("Not Eligible For Voting in India")
```

```python
# CHECK WHETHER A PERSON IF HE IS ELIGIBLE FOR BUS PASS DISCOPUNT IN MAHARASHATRA
gender=input("Enter Your Gender (male/female) : ")
state=input("Are You From Maharashatra? (yes/no) : ")
if gender=="female" and state=="yes" :
  print("You Are Eligible For Bus Pass Discount")
elif gender=="female" and state=="no" :
  print("You Are Not From Maharashatra so you are not eligible")
elif gender=="male" and state=="no" :
  print("You Are male and also you are not from Maharashatra")
else:
  print("Your Are male so you are not eligible")
```

```python
# NESTED IF 2)
marks = int(input("Enter the student's marks: "))

if marks >= 0 and marks <= 100:
    if marks >= 90:
        grade = "A"
    else:
        if marks >= 80:
            grade = "B"
        else:
            if marks >= 70:
                grade = "C"
            else:
                if marks >= 60:
                    grade = "D"
                else:
                    grade = "F"

    print(f"The student's grade is: {grade}")
else:
    print("Invalid marks! Please enter a value between 0 and 100.")
```
